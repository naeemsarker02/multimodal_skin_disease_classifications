## 5. Image dimensions (sampled 250)

Highly heterogeneous smartphone-captured images: width/height range from 167px to 3120px, almost always square-cropped (aspect ratio ~1.0, max 1.16). File sizes range 44KB-12.4MB. This heterogeneity — not a fixed capture protocol like HAM10000 — should directly inform the Phase 6 input-resolution/resize choice: a single resize target (e.g. 224x224 or 256x256) is unavoidable and will discard real resolution information for the largest images.
